```python
# 4. Iniciar o servidor Ollama
import subprocess
import time

processo_ollama = subprocess.Popen(
    ["ollama", "serve"],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)

time.sleep(5)

print("Servidor Ollama iniciado.")
```

```python
# 5. Baixar o modelo
!ollama pull llama3.2:1b
```

```python
from langchain_ollama import OllamaLLM

def processamento_resposta(pergunta):
    llm = OllamaLLM(
        model="llama3.2:1b",
        num_predict=1024
    )

    resposta = llm.invoke(pergunta)

    return resposta
```

```python
pergunta = "Explique GoLang de forma simples para um iniciante."

resposta = processamento_resposta(pergunta)

print(resposta)
```
